```python
import os
import pandas as pd
folder_path = r"C:\Users\MainUser\Downloads\cs230_data_full"
files = [
    f for f in os.listdir(folder_path)
    if f.endswith(".zst") and os.path.isfile(os.path.join(folder_path, f))
]
print(len(files))
```

```python
def process_data(daily_df):

    daily_df["weighted_mid"] = (
    daily_df["bid_px_00"] * daily_df["ask_sz_00"] + daily_df["ask_px_00"] * daily_df["bid_sz_00"]
        ) / (daily_df["bid_sz_00"] + daily_df["ask_sz_00"])

    daily_df["weighted_mid"] = daily_df["weighted_mid"].ffill()
    daily_df = daily_df.sort_index()
    daily_df = daily_df.resample("1s").ffill()

    daily_df["ret_1s"] = daily_df["weighted_mid"].pct_change()
    daily_df["ret_sq"] = daily_df["ret_1s"] ** 2
    rv_60s = daily_df["ret_sq"].resample("60s", label="right", closed="right").sum()

    daily_df["rv_60s"] = rv_60s.reindex(daily_df.index, method="ffill")
    ret_1min = daily_df["weighted_mid"].resample("60s", label="right", closed="right").last().pct_change()
    daily_df["ret_60s"] = ret_1min.reindex(daily_df.index, method="ffill")
    first_date = daily_df.index[0].normalize()
    session_start = first_date + pd.Timedelta(hours=8, minutes=1)
    daily_df = daily_df[daily_df.index >= session_start]
    df_1min = daily_df[daily_df.index.second == 0].copy()
    # print(df_1min[["rv_60s", "ret_60s", "weighted_mid"]].head(10))
    return df_1min
```

```python
#%pip install databento
import databento as db

main_folder = r"C:\Users\MainUser\cs230\BernsteinMartingaleNet\MarketData\historical_data"

for file_name in files:

    #if "20250220" not in file_name:
    #    continue
    sample_file = f"{folder_path}\\{file_name}"
    data = db.DBNStore.from_file(sample_file)
    df_1min = process_data(data.to_df())
    df_1min = df_1min.reset_index()
    df_1min = df_1min[["ts_recv", "price", "bid_px_00", "ask_px_00", "bid_sz_00", "ask_sz_00",
                        "weighted_mid", "rv_60s", "ret_60s"]]
    df_1min.set_index("ts_recv", inplace=True)
    date_str = file_name.split("-")[2].split(".")[0]
    df_1min.to_csv(os.path.join(main_folder, f"spy_1min_{date_str}.csv"))
    print(date_str, df_1min.shape)
```
