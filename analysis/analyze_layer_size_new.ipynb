```python
row_to_color = "dev"   # <- change this to which row you want

def heatmap_single_row(row):
    """Apply background color only to the selected row."""
    if row.name != row_to_color:
        # return no styling for other rows
        return ['' for _ in row]

    # scale the row values for heatmap (0–1 range)
    vals = row.values.astype(float)
    norm = (vals - vals.min()) / (vals.max() - vals.min() + 1e-12)

    # convert normalized values to colors using a colormap
    import matplotlib.cm as cm
    cmap = cm.get_cmap("viridis")

    return [
        f"background-color: rgba{cmap(v)}"
        for v in norm
    ]
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import HTML
from pathlib import Path

# ============================================================
# Custom Green colormap (light → dark)
# ============================================================
green_cmap = plt.get_cmap("Greens")   # safe, never deprecated


# ============================================================
# Heatmap ONLY on the "dev" row
# ============================================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

green_cmap = plt.get_cmap("Greens")

def heatmap_dev_row(row):
    if row.name != "dev":
        return ['' for _ in row]

    vals = row.values.astype(float)
    norm = (vals - vals.min()) / (vals.max() - vals.min() + 1e-12)

    css_list = []
    for v in norm:
        # RGBA from colormap
        r, g, b, a = green_cmap(v)
        R, G, B = int(r * 255), int(g * 255), int(b * 255)

        # Luminance (perceptual brightness)
        luminance = 0.299 * R + 0.587 * G + 0.114 * B

        # Choose text color based on brightness
        text_color = "black" if luminance > 140 else "white"

        css_list.append(
            f"background-color: rgba({R},{G},{B},{a}); color: {text_color}"
        )

    return css_list



# ============================================================
# Your settings
# ============================================================


distribution_heads = ["BLogistic"]
model_specs = ["NoReg", "NoReg_RV", "NoReg_RV_Decay"]

def layer_to_str(layer):
    return "-".join(map(str, layer))


results = {}

# ============================================================
# Main Loop with Styled Display
# ============================================================
for head in distribution_heads:
    for spec in model_specs:
        table = pd.DataFrame()

        for layer in layer_sizes:
            layer_str = layer_to_str(layer)
            file_path = base_path / f"{head}{spec}_{layer_str}_{dof}" / "final_losses.csv"

            if not file_path.exists():
                continue

            df = pd.read_csv(file_path)
            table[layer_str] = df.set_index("data_type")["loss"]

        table.index.name = None
        table = table.round(4)
        results[(head, spec)] = table

        print(f"\n=====Synthetic Data ===== {head} {spec} DoF-{dof} =====\n")

        styled = (
            table.style
                .format("{:.4f}")
                .apply(heatmap_dev_row, axis=1)    # 🔥 Green heatmap on dev row only
                .set_table_styles([
                    {'selector': 'table', 'props': [('font-size', '18px')]},
                    {'selector': 'th',    'props': [('font-size', '18px')]},
                    {'selector': 'td',    'props': [('font-size', '18px')]},
                ])
        )

        display(HTML(styled.to_html()))
```
